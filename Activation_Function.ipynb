```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)

sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)
leaky_relu = np.where(x > 0, x, 0.01 * x)
elu = np.where(x > 0, x, np.exp(x) - 1)
prelu = np.where(x > 0, x, 0.25 * x)
softplus = np.log(1 + np.exp(x))

exp_x = np.exp(x - np.max(x))     
softmax = exp_x / np.sum(exp_x)

plt.figure(figsize=(12, 8))
plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, tanh, label="Tanh")
plt.plot(x, relu, label="ReLU")
plt.plot(x, leaky_relu, label="Leaky ReLU")
plt.plot(x, elu, label="ELU")
plt.plot(x, prelu, label="PReLU")
plt.plot(x, softplus, label="Softplus")
plt.plot(x, softmax, label="Softmax")

plt.title("Activation Functions (Library Implementations)")
plt.xlabel("Input x")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def sigmoid_manual(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid_manual(x)
    return s * (1 - s)

y = sigmoid_manual(x)
dy = sigmoid_derivative(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Sigmoid")
plt.plot(x, dy, label="Sigmoid Derivative")
plt.title("Sigmoid Activation and Derivative")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def tanh_manual(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

def tanh_derivative(x):
    return 1 - tanh_manual(x) ** 2

y = tanh_manual(x)
dy = tanh_derivative(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Tanh")
plt.plot(x, dy, label="Tanh Derivative")
plt.title("Tanh Activation and Derivative")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def relu_manual(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)  

y = relu_manual(x)
dy = relu_derivative(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="ReLU")
plt.plot(x, dy, label="ReLU Derivative")
plt.title("ReLU Activation and Derivative")
plt.legend()
plt.grid(True)
plt.show()
```
