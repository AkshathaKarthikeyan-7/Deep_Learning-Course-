```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(-5, 5, 100)
noise = np.random.normal(0, 0.5, size=len(x))
y_true = 3 * x + 2 + noise

offsets = np.linspace(-5, 5, 200)

mae_list = []
rmse_list = []

for offset in offsets:
    y_pred = y_true + offset
    mae = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mae_list.append(mae)
    rmse_list.append(rmse)

plt.figure(figsize=(8, 5))
plt.plot(offsets, mae_list, label="MAE")
plt.plot(offsets, rmse_list, label="RMSE")
plt.xlabel("Prediction Error (Offset)")
plt.ylabel("Loss")
plt.title("MAE and RMSE vs Prediction Error")
plt.legend()
plt.grid(True)
plt.show()
```

```python
y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.1, 0.4, 0.8])

bce_manual = -np.mean(
    y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
)

print("Binary Cross-Entropy (Manual):", bce_manual)
```

```python
y_true = np.array([[1,0,0], [0,1,0]])
y_pred = np.array([[0.7,0.2,0.1], [0.1,0.8,0.1]])

cce_manual = -np.mean(np.sum(y_true * np.log(y_pred), axis=1))

print("Categorical Cross-Entropy (Manual):", cce_manual)
```

```python
y_true = np.array([0, 1])
y_pred = np.array([[0.7,0.2,0.1], [0.1,0.8,0.1]])

sce_manual = -np.mean(np.log(y_pred[np.arange(len(y_true)), y_true]))

print("Sparse Categorical Cross-Entropy (Manual):", sce_manual)
```

```python
scores = np.linspace(-3, 3, 400)
y = 1  

hinge_loss = np.maximum(0, 1 - y * scores)
squared_hinge_loss = hinge_loss ** 2

plt.figure(figsize=(8, 5))
plt.plot(scores, hinge_loss, label="Hinge Loss")
plt.plot(scores, squared_hinge_loss, label="Squared Hinge Loss")
plt.xlabel("Prediction Confidence f(x)")
plt.ylabel("Loss")
plt.title("Hinge vs Squared Hinge Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
P = np.array([0.7, 0.2, 0.1])

eps = np.linspace(0.01, 0.6, 100)
kl_values = []

for e in eps:
    Q = np.array([0.7 - e, 0.2 + e/2, 0.1 + e/2])
    kl = np.sum(P * np.log(P / Q))
    kl_values.append(kl)

plt.figure(figsize=(8, 5))
plt.plot(eps, kl_values)
plt.xlabel("Deviation from Target Distribution")
plt.ylabel("KL Divergence")
plt.title("KL Divergence vs Prediction Deviation")
plt.grid(True)
plt.show()
```
